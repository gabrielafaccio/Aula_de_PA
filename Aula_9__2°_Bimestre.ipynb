```python
# Loops  - laços e oterações

for i in range (5) :
  print ("Olá Mundo!")
```

```python
# Escrever um programa que recebe um número e retorno somente os números pares


numero = int (input("Digite um Número: "))

for i in range (numero + 1) :
  if i % 2  == 0:
      print (i)
```

```python
# Escrever um programa que recebe um número e imprime uma contagem regressiva

numero = int (input("Digite um número: "))

for i in range (numero + 1) :
  print(numero - i)
```

```python
# Escrever um programa que recebe um número de 1 a 10 e imprime a tabuada

numero = int (input("Digite um número de 1 a 10: "))


for i in range(1, 11):
  print(f"{numero} x {i} = {numero * i}")
```
