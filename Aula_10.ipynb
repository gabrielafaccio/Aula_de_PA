```python
# 1 - Contagem simples, o úsuario digita um número e o programa imprime do zero até o número]

# ENTRADA

numero = int (input("Digite um Número: "))

# PROCESSAMENTO/SAÍDA

for i in range (numero + 1) :
  print  (i)
```

```python
# 2 - Contagem progessiva, o úsuario digita um número e o programa  imprime do número até o zero


# ENTRADA

numero = int (input("Digite um Número: "))

# PROCESSAMENTO/SAÍDA

for i in range (numero + 1) :
  print(numero - i)






```

```python
# 3 - Números pares, o úsuario digita um número e o programa imprime os números pares

# ENTRADA

numero = int (input("Digite um Número: "))

# PROCESSAMENTO/ SAÍDA

for i in range (numero + 1) :
  if i % 2  == 0:
      print (i)


```

```python
# 4 - Números ímpares, o Usuario igita um número e o programa imprime só os números ímpares


# ENTRADA

numero = int (input("Digite um Número: "))

# PROCESSAMENTO/ SAÍDA

for i in range (numero + 1) :
  if i % 2  == 1:
      print (i)
```

```python
# 5 - Soma, o usuario digita um número e o programa soma os números do zero até o número

# ENTRADA

numero = int (input("Digite um número de 1 a 10: "))

# PROCESSAMENTO/SAÍDA

for i in range(0,7):
  print(f"{numero} + {i} = {numero + i}")





```

```python
# 6 - Tabuada, o usúario digita um número e o programa imprime a tabuada do número

# ENTRADA

numero = int (input("Digite um número de 1 a 10: "))

# PROCESSAMENTO/SAÍDA

for i in range(0, 11):
  print(f"{numero} X {i} = {numero * i}")
```

```python
# 7 - Contar letras, o usúario digita um palavra e o Programa conta as letras da palavra

# ENTRADA
palavra = input("Digite uma palavra: ")

# PROCESSAMENTO/SAÍDA
letras = len(palavra)
print(f"A palavra '{palavra}' tem {letras} letras.")
```

```python
# 8 - Contar vogais, o usúario digita uma palavra e o programa conta as vogais

# ENTRADA
palavra = input("Digite uma palavra: ").lower()

# PROCESSAMENTO/SAÍDA
vogais = "aeiou"
contador_vogais = 0
for letra in palavra:
  if letra in vogais:
    contador_vogais += 1

print(f"A palavra '{palavra}' tem {contador_vogais} vogais.")
```

```python
# 9 - Maior número digitado, o usúario digita quatro números e o programa encontra o maior entre eles

# ENTRADA
print("Digite quatro números:")
numero1 = int(input("Digite o Número 1: "))
numero2 = int(input("Digite o Número 2: "))
numero3 = int(input("Digite o Número 3: "))
numero4 = int(input("Digite o Número 4: "))

# PROCESSAMENTO/SAÍDA
maior_numero = numero1


if numero2 > maior_numero:
  maior_numero = numero2
if numero3 > maior_numero:
  maior_numero = numero3
if numero4 > maior_numero:
  maior_numero = numero4

print(f"O maior número digitado é: {maior_numero}")
```

```python
# 10 - O usúario digita um número entre 01 e 100 e o programa mostra quais são divisíveis por 3 e 5

# ENTRADA

numero = int(input("Digite um número entre 1 e 100: "))

# PROCESSAMENTO/SAÍDA

print(f"Números divisíveis por 3 e 5 até {numero}:")
for i in range(1, numero + 1):
  if i % 3 == 0 and i % 5 == 0:
    print(i)
```
