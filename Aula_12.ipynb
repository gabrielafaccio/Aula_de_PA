```python
# Escrever um jogo de adivinhação o uusuario tera 3 chances para adivinhar um número de 0 a 10
# O programa dar uma mensagem se o palpite foi maior ou menor
# O programa avisar quente ou frio # Regra extra tem que ser número inteiro


import random

numero_secreto = random.randint(0, 10)
chances = 3

print("Bem-vindo ao jogo de adivinhação!")
print("Estou pensando em um número inteiro entre 0 e 10. Você tem 3 chances para adivinhar.")

jogando = True
for tentativa in range(1, chances + 1):
    palpite = int(input(f"Chance {tentativa}/{chances}. Qual é o seu palpite? "))

    if palpite == numero_secreto:
        print(f"Parabéns! Você adivinhou o número {numero_secreto} em {tentativa} tentativas!")
        jogando = False
        break
    else:
        if palpite < numero_secreto:
            print("Seu palpite é muito baixo.")
        else:
            print("Seu palpite é muito alto.")

        diferenca = abs(palpite - numero_secreto)
        if diferenca <= 2:
            print("Você está quente!")
        else:
            print("Você está frio.")

if jogando:
    print(f"Suas chances acabaram! O número secreto era {numero_secreto}.")
```
