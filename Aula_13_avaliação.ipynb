QUESTÃO 2

```python
# Solicite ao usuário um número inteiro positivo. b Utilizando um laço de repetição, calcule
# e exiba a soma de  todos os números de 1 até o número informado.


numero = int(input("Digite um número: "))

soma = 0

for i in range(1, numero + 1):
    soma += i

print(f"Resultado: {soma}")
```

```python
# Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

numero = int(input("Digite um número: "))

multiplicar = numero

for i in range(1, numero * 1):

    multiplicar *= i

print(f"Resultado: {multiplicar}")
```

```python
# Um caixa eletrônico possui notas de:

# R$ 100
# R$ 50
# R$ 20
# R$ 10
# R$ 5
# R$ 2
# R$ 1

#Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.


valor = int(input("Digite o valor do saque: "))

notas = [100, 50, 20, 10, 5, 2, 1]

for nota in notas:
  if valor >= nota:
    qtd = valor // nota
    print(f"{qtd} nota(s) de R$ {nota}")
    valor %= nota

if valor > 0:
  print("Não foi possível sacar o valor exato com as notas disponíveis.")
```
