```python
# escrever um prgrama que recebe as notas do aluno rteferente aos 4 bimestres.

# Calcular a média das motas e imprimir a mensagem de aprovado, recuperação ou aprovado

# nota maior ou igual a 7 aprovado
# nota maior ou igual a 5 recuperação
# nota menor que 5 reprovado

# Entrada

nota_1 = float(input ("Digite a nota do 1° Bimestre "))
if nota_1 < 0 or nota_1 > 10:
  print("nota inválida, digite uma nota de 0 a 10:")


nota_2 = float(input ("Digite a nota do 2° Bimestre "))
if nota_2 < 0 or nota_2 > 10:
   print("nota inválida, digite uma nota de 0 a 10:")


nota_3 = float(input ("Digite a nota do 3° Bimestre "))
if nota_3 < 0 or nota_3 > 10:
   print("nota inválida, digite uma nota de 0 a 10:")

nota_4 = float(input ("Digite a nota do 4° Bimestre "))
if nota_4 < 0 or nota_4 > 10:
   print("nota inválida, digite uma nota de 0 a 10:")

# Processamento/Saída

media = (nota_1 + nota_2 + nota_3 + nota_4) /4

if media < 5:
    print(f"Você está reprovado, sua media é :{media} ")

elif media <= 7:
    print(f"Você está de recuperação, sua mediaa é {media}")

else:
    print(f"Você está aprovado, sua media é {media}")
```
