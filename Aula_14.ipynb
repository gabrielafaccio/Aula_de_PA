Exercício 1 — Contagem

```python
numero = int(input("Digite um número: "))

i = 1

while i  <= numero:
  print(i)
  i += 1
```

Exercício 2 — Contagem regressiva

```python
numero = int(input("Digite um número: "))

while numero >= 0:
    print(numero)
    numero -= 1
```

Exercício 3 — Números pares

```python
numero = 0

while numero <= 20:
    print (numero)
    numero += 2
```

Exercício 4 — Soma dos números

```python
numero = int(input("Digite um número: "))

i = 1
soma = 0

while i <= numero:
    soma += i
    i += 1

print ("Soma:" , soma)
```

Exercício 5 — Tabuada

```python
numero = int(input("Digite um número:"))

i = 1

while i <= 10:
    print (numero, "x", i, "=", numero * i)
    i += 1
```

Exercício 7 — Somando até digitar zero

```python
soma = 0

numero = int(input("Digite um número: "))

while numero != 0:
  soma += numero
  numero = int(input("Digite um número: "))

print ("Soma:", soma)
```

Exercício 8 — Maior número

```python
maior = 0
numero = int(input("Digite um número: "))

while numero != 0:
  if numero > maior:
      maior = numero
  numero = int(input("Digite um número: "))

print(f"Maior número: {maior}")
```

Exercício 9 — Média das notas

```python
soma = 0
contador = 1

while contador <= 5:
    nota = float(input(f"Digite a nota {contador}: "))
    soma += nota
    contador += 1

media = soma / 5
print(f"Média: {media}")
```
